# Day 13 - all_estimators Model Comparison


## 학습 목표
- sklearn all_estimators로 회귀/분류 모델 일괄 비교
- 데이터셋별 성능 정리


## 1. 회귀 모델 일괄 비교 (Diabetes)


```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_diabetes, load_breast_cancer
from sklearn.model_selection import cross_val_score, KFold
from sklearn.preprocessing import StandardScaler
from sklearn.utils import all_estimators
import warnings
warnings.filterwarnings('ignore')

data = load_diabetes()
X, y = data.data, data.target
X = StandardScaler().fit_transform(X)

regressors = all_estimators(type_filter='regressor')
results = []
kfold = KFold(n_splits=3, shuffle=True, random_state=42)

for name, RegClass in regressors:
    try:
        model = RegClass()
        scores = cross_val_score(model, X, y, cv=kfold, scoring='r2', n_jobs=-1)
        results.append({'Model': name, 'R2_mean': scores.mean(), 'R2_std': scores.std()})
    except Exception:
        pass

df = pd.DataFrame(results).sort_values('R2_mean', ascending=False)
print("=== Top 10 Regressors (Diabetes) ===")
print(df.head(10).to_string(index=False))
```

## 2. 분류 모델 일괄 비교 (Breast Cancer)


```python
data = load_breast_cancer()
X, y = data.data, data.target
X = StandardScaler().fit_transform(X)

classifiers = all_estimators(type_filter='classifier')
results_clf = []
kfold = KFold(n_splits=3, shuffle=True, random_state=42)

for name, ClfClass in classifiers:
    try:
        model = ClfClass()
        scores = cross_val_score(model, X, y, cv=kfold, scoring='accuracy', n_jobs=-1)
        results_clf.append({'Model': name, 'Acc_mean': scores.mean(), 'Acc_std': scores.std()})
    except Exception:
        pass

df_clf = pd.DataFrame(results_clf).sort_values('Acc_mean', ascending=False)
print("=== Top 10 Classifiers (Breast Cancer) ===")
print(df_clf.head(10).to_string(index=False))
```
