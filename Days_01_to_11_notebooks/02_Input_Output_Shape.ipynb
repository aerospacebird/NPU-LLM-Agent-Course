# Day 2 - Input/Output Data Structure & Shape Check


## 1. 데이터 생성 및 shape 확인


```python
import numpy as np
import tensorflow as tf

np.random.seed(42)
X = np.random.rand(1000, 8).astype(np.float32)   # (samples, features)
y = np.random.rand(1000, 1).astype(np.float32)

print("X shape :", X.shape)
print("y shape :", y.shape)
print("X dtype :", X.dtype)
print("X min/max:", X.min(), X.max())
```

## 2. 모델 입출력 shape 확인


```python
model = tf.keras.Sequential([
    tf.keras.layers.Dense(64, activation='relu', input_shape=(8,)),
    tf.keras.layers.Dense(32, activation='relu'),
    tf.keras.layers.Dense(1)
])

print("Model input_shape :", model.input_shape)
print("Model output_shape:", model.output_shape)
model.summary()
```

## 3. 예측 결과 shape 확인


```python
pred = model.predict(X[:5], verbose=0)
print("Prediction shape:", pred.shape)
print("Prediction values:\n", pred)
```
